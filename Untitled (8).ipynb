```python

funds_net = pd.read_excel("project Funds_2026.xlsx", sheet_name="Net")
rm = factors["RF"] + factors["MKT"]
rm_cml = rm.mean() * 12
rm_std = rm.std() * np.sqrt(12)

slope = (rm_cml - rf_annual) / rm_std

def cml(df, title, drop_cols=("Year", "Month")):
    cols = df.select_dtypes("number").columns.drop(list(drop_cols), errors="ignore")
    means = df[cols].mean() * 12
    sigma = df[cols].std() * np.sqrt(12)
    
    plt.figure(figsize=(10, 5), dpi=120)
    plt.scatter(sigma*100, means*100, s=18, alpha=0.7)

    x = np.linspace(0, sigma.max() * 1.1, 200)
    y = rf_annual + slope * x

    plt.plot(x * 100, y * 100, lw=2, color="purple", label="CML")
    plt.scatter([rm_std * 100], [rm_cml * 100], marker="*", s=180, label="Market")

    plt.grid(alpha=0.25)
    plt.title(title)
    plt.xlabel("Annualized standard deviations (%)")
    plt.ylabel("Annualized mean (%)")
    plt.legend(frameon=False)
    plt.tight_layout()
    plt.show()

cml(returns, "Individual Stocks")
cml(funds_net, "Funds(net)")
```

Individual stocks vary a lot: some have high returns, some really low, and many are super volatile without giving higher returns. Many of them fall below the Capital Market Line, meaning their risk isn’t really being rewarded. Some extreme cases have huge volatility and high returns, but that is mostly idiosyncratic risk, not market risk. 
However, funds are more consistent, less volatile, and many sit closer to, or even above the CML. This shows that diversification works: it reduces risk and improves returns. Overall, these patterns support what we learn in class: only systematic risk gets rewarded, and diversification makes your portfolio more efficient.



# OUESTION 3

```python
idx = returns.index.intersection(factors.index)
ret = returns.loc[idx]
fac = factors.loc[idx]

stock_columns = ret.select_dtypes("number").columns.drop(["Year", "Month"], errors="ignore")
rf_annual_idx = fac["RF"].mean() * 12
rm_annual_idx = fac["MKT"].mean() * 12
smb_annual_idx = fac["SMB"].mean() * 12
hml_annual_idx = fac["HML"].mean() * 12

def ols_res(y, X):
    X = sm.add_constant(X, has_constant="add")
    return sm.OLS(y, X).fit()

res = pd.DataFrame(index=stock_columns, columns=["beta_capm_market", "beta_ff3_market", "beta_smb", "beta_hml", "nobs"], dtype=float)

for i in stock_columns:
    y = (ret[i] - fac["RF"]).rename("excess")

    df_capm = pd.concat([y, fac[["MKT"]]], axis=1).dropna()
    if df_capm.shape[0] < 12:
        continue
    capm = ols_res(df_capm["excess"], df_capm[["MKT"]])

    df_ff3 = pd.concat([y, fac[["MKT", "SMB", "HML"]]], axis=1).dropna()
    if df_ff3.shape[0] < 12:
        continue
    ff3 = ols_res(df_ff3["excess"], df_ff3[["MKT", "SMB", "HML"]])

    res.loc[i, "beta_capm_market"] = capm.params["MKT"]
    res.loc[i, "beta_ff3_market"] = ff3.params["MKT"]
    res.loc[i, "beta_smb"] = ff3.params["SMB"]
    res.loc[i, "beta_hml"] = ff3.params["HML"]
    res.loc[i, "nobs"] = ff3.nobs
    
res = res.dropna()

res["capm_annual"] = rf_annual_idx + res["beta_capm_market"] * rm_annual_idx
res["ff3_annual"] = rf_annual_idx + res["beta_ff3_market"] * rm_annual_idx + res["beta_smb"]* smb_annual_idx + res["beta_hml"]*hml_annual_idx
res["diff"] = (res["ff3_annual"] - res["capm_annual"]) * 100

outs = res[["beta_capm_market", "beta_ff3_market", "beta_smb", "beta_hml", "capm_annual", "ff3_annual", "diff"]].copy()
outs["capm_annual"] *=100 
outs["ff3_annual"] *=100
display(outs.round(4))
market_cap_mean = market_cap.select_dtypes("number").mean()
res_size = res.join(market_cap_mean.rename("cap_mean"), how="inner")

cap_median = res_size["cap_mean"].median()
res_size["size_group"] = np.where(res_size["cap_mean"] <= cap_median, "Small", "Large")

summary = res_size.groupby("size_group")[["capm_annual", "ff3_annual"]].mean() * 100
summary["FF3 - CAPM"] = (res_size.groupby("size_group")["ff3_annual"].mean() - res_size.groupby("size_group")["capm_annual"].mean()) * 100


display(summary.round(4))
```

Yes, the predictions differ between the two models. Overall, the differences are small, but they are not consistent across firm size. For large stocks, CAPM and the Fama–French three-factor model give almost identical expected returns, indicating that additional risk factors add little explanatory power. For small stocks, however, the three-factor model predicts higher expected returns than CAPM. This suggests that CAPM fails to capture the size-related risk premium, which is explicitly accounted for in the Fama–French model.

```python
start_date = "2016-01"
min_obs_alpha = 24 

def prepare_df(df, date_col="Date"):
    df = df.copy()
    if date_col in df.columns:
        df[date_col] = pd.to_datetime(df[date_col], errors="coerce")
        df = df.dropna(subset=[date_col]).set_index(date_col)
    df.index = pd.to_datetime(df.index, errors="coerce")
    # drop Year/Month columns if exist
    for c in ["Year","Month","year","month"]:
        if c in df.columns:
            df.drop(columns=[c], inplace=True)
    return df

ret = prepare_df(returns)
mcap = prepare_df(market_cap)
f = prepare_df(factors)[["RF","MKT"]]

start_ts = pd.to_datetime(start_date + "-01")
ret = ret.loc[ret.index >= start_ts]
mcap = mcap.loc[mcap.index >= start_ts]
f = f.loc[f.index >= start_ts]

common_dates = ret.index.intersection(mcap.index).intersection(f.index)
ret = ret.loc[common_dates]
mcap = mcap.loc[common_dates]
f = f.loc[common_dates]

# convert to PeriodIndex for monthly data
ret.index = ret.index.to_period("M")
mcap.index = mcap.index.to_period("M")
f.index = f.index.to_period("M")

def build_portfolios(ret, mcap):
    portfolios = []
    for y in sorted(ret.index.year.unique()):
        months_y = ret.index[ret.index.year == y]
        if len(months_y) == 0:
            continue
        first_m = months_y.min()
        cap0 = mcap.loc[first_m].dropna()
        eligible = cap0.index.intersection(ret.columns)
        cap0 = cap0.loc[eligible].dropna()
        if len(cap0) < 5:
            continue
        # assign quintiles
        q = pd.qcut(cap0.rank(method="first"), 5, labels=[1,2,3,4,5], duplicates="drop")
        groups = {int(k): list(q.index[q == k]) for k in q.cat.categories}
        # compute portfolio returns
        for m in months_y:
            r_m = ret.loc[m]
            row = {"Date": m}
            for k in sorted(groups.keys()):
                vals = r_m[groups[k]].dropna()
                row[f"Q{k}"] = vals.mean() if len(vals)>0 else np.nan
            portfolios.append(row)
    port_ret = pd.DataFrame(portfolios).set_index("Date").sort_index()
    port_cols = [c for c in ["Q1","Q2","Q3","Q4","Q5"] if c in port_ret.columns]
    return port_ret[port_cols]

port_ret = build_portfolios(ret, mcap)
# rename quintiles Q1–Q5 → P1–P5
port_ret = port_ret.rename(columns={"Q1":"P1", "Q2":"P2", "Q3":"P3", "Q4":"P4", "Q5":"P5"})

summary_rows = []

for p in port_ret.columns:
    r_p = port_ret[p]
    # monthly/annualized stats
    mean_m = r_p.mean()
    std_m = r_p.std()
    mean_a = mean_m * 12
    std_a = std_m * np.sqrt(12)
    # CAPM regression
    df_reg = pd.concat([(r_p - f["RF"]), f["MKT"]], axis=1).dropna()
    df_reg.columns = ["excess_port","MKT"]
    if len(df_reg) < min_obs_alpha:
        alpha_m = alpha_a = beta_mkt = tval = pval = R2 = np.nan
        sig = False
    else:
        X = sm.add_constant(df_reg["MKT"])
        res = sm.OLS(df_reg["excess_port"], X).fit()
        alpha_m = res.params["const"]
        alpha_a = alpha_m * 12
        beta_mkt = res.params["MKT"]
        tval = res.tvalues["const"]
        pval = res.pvalues["const"]
        R2 = res.rsquared
        sig = pval < 0.05
    summary_rows.append({
        "Portfolio": p,
        "mean_monthly": mean_m,
        "std_monthly": std_m,
        "mean_annual": mean_a,
        "std_annual": std_a,
        "CAPM_alpha_monthly": alpha_m,
        "CAPM_alpha_annual": alpha_a,
        "CAPM_alpha_t": tval,
        "CAPM_alpha_p": pval,
        "CAPM_alpha_sig_5pct": sig,
        "CAPM_beta": beta_mkt,
        "CAPM_R2": R2
    })

summary = pd.DataFrame(summary_rows).set_index("Portfolio").round(6)
display(summary)
```

Using post-2015 data, stocks are sorted each year by beginning-of-year market capitalization into five equally weighted portfolios (Q1 = smallest, Q5 = largest). The results show a clear size pattern in returns: average returns generally decline as firm size increases. However, Q2 has the highest average return and also the highest volatility, followed closely by Q1. Larger portfolios have both lower average returns and lower volatility.

The CAPM regressions further show that Q1 and Q2 have statistically significant positive alpha at the 5% level, while Q3 to Q5 do not , although Q3 is at the margin.This means that for the smallest portfolios, returns are higher than what can be explained by market beta alone. In other words, CAPM underpredicts returns for small stocks. By contrast, the alphas for large stocks are not significantly different from zero, indicating that their returns are largely explained by market risk.
Overall, the evidence suggests a size effect: small stocks earn excess returns that CAPM cannot fully explain, whereas large stocks are consistent with CAPM predictions.

```python
import numpy as np
import pandas as pd
import statsmodels.api as sm

min_obs = 24
start_date = "2016-01-01"  # filtrerar från och med denna månad

# ---------- helpers ----------
def to_period(df, date_col=None):
    df = df.copy()
    if date_col:
        df[date_col] = pd.to_datetime(df[date_col], errors="coerce")
        df = df.dropna(subset=[date_col]).set_index(date_col)
    else:
        df.index = pd.to_datetime(df.index, errors="coerce")
    df.index = df.index.to_period("M")
    return df.sort_index()

def comma_to_float(df):
    df = df.copy()
    for c in df.columns:
        df[c] = pd.to_numeric(
            df[c].astype(str).str.replace(",", ".", regex=False),
            errors="coerce"
        )
    return df

def load_fund(sheet_name, funds_path):
    df = pd.read_excel(funds_path, sheet_name=sheet_name)
    df.columns = df.columns.str.strip()

    # om Year/Month finns: skapa riktig Date från dem (robust)
    if {"Year", "Month"}.issubset(df.columns):
        df["Date"] = pd.to_datetime(
            df["Year"].astype(int).astype(str) + "-" +
            df["Month"].astype(int).astype(str).str.zfill(2) + "-01",
            errors="coerce"
        )
        df = df.drop(columns=["Year", "Month"], errors="ignore")

    # sätt PeriodIndex (M)
    df = to_period(df, date_col="Date" if "Date" in df.columns else None)

    # se till att allt som kan vara tal blir tal
    df = comma_to_float(df).dropna(axis=1, how="all")
    return df

def load_fund_types_auto(funds_path):
    # Funds-filen har en "Info"-sheet med Fund name och fund type
    info = pd.read_excel(funds_path, sheet_name="Info")
    info.columns = info.columns.str.strip()

    # normalisera kolumnnamn lite
    name_col = None
    type_col = None
    for c in info.columns:
        cl = c.strip().lower()
        if cl in ["fund name", "fund_name", "name"]:
            name_col = c
        if cl in ["fund type", "fund_type", "type"]:
            type_col = c

    if name_col is None or type_col is None:
        # om något oväntat i filen
        return pd.DataFrame(columns=["Type"]).astype({"Type": "string"})

    out = info[[name_col, type_col]].copy()
    out.columns = ["Fund", "Type"]
    out["Fund"] = out["Fund"].astype(str).str.strip()
    out["Type"] = out["Type"].astype(str).str.strip()
    return out.set_index("Fund")

def type_bucket(x):
    x = str(x).lower()
    if "index" in x:
        return "Index"
    if "active" in x or "actively" in x:
        return "Active"
    if x in ["nan", "none", ""]:
        return np.nan
    return "Other"

def capm(funds, factors, label="net"):
    # OBS: factors-kolumner i din fil: rf och rm_rf
    X = sm.add_constant(factors["rm_rf"].rename("rm_rf"))
    rows = []

    for fund in funds.columns:
        y = (funds[fund] - factors["rf"]).rename("excess")
        tmp = pd.concat([y, X], axis=1).dropna()

        if len(tmp) < min_obs:
            continue

        fit = sm.OLS(tmp["excess"], tmp[["const", "rm_rf"]]).fit()

        rows.append({
            "Fund": fund,
            f"alpha_{label}_monthly": fit.params.get("const", np.nan),
            f"alpha_{label}_annual":  fit.params.get("const", np.nan) * 12,
            f"alpha_{label}_t":       fit.tvalues.get("const", np.nan),
            f"alpha_{label}_p":       fit.pvalues.get("const", np.nan),
            f"beta_{label}":          fit.params.get("rm_rf", np.nan),
            f"R2_{label}":            fit.rsquared,
            "N": int(fit.nobs)
        })

    if len(rows) == 0:
        return pd.DataFrame(columns=["alpha","beta"]).rename_axis("Fund")

    return pd.DataFrame(rows).set_index("Fund")


# ---------- paths (bara filnamn om du har dem i samma mapp) ----------
funds_path = "project Funds_2026.xlsx"
factors_path = "project Factors_2026.xlsx"

# ---------- load funds ----------
funds_net   = load_fund("Net", funds_path)
funds_gross = load_fund("Gross", funds_path)

# ---------- load factors (rf, rm_rf) ----------
factors_raw = pd.read_excel(factors_path)
factors_raw.columns = factors_raw.columns.str.strip()

# använd Date-kolumnen i factors-filen
factors_subset = factors_raw[["Date", "rf", "rm_rf"]].copy()
factors_subset = to_period(factors_subset, date_col="Date").dropna()

# ---------- filter start_date ----------
start_p = pd.Period(pd.to_datetime(start_date), "M")
funds_net   = funds_net.loc[funds_net.index >= start_p]
funds_gross = funds_gross.loc[funds_gross.index >= start_p]
factors_subset = factors_subset.loc[factors_subset.index >= start_p]

# ---------- common months ----------
common_months = funds_net.index.intersection(funds_gross.index).intersection(factors_subset.index)

funds_net = funds_net.loc[common_months]
funds_gross = funds_gross.loc[common_months]
factors_subset = factors_subset.loc[common_months]

# ---------- run CAPM ----------
alpha_net   = capm(funds_net, factors_subset, "net")
alpha_gross = capm(funds_gross, factors_subset, "gross")

both = alpha_net.index.intersection(alpha_gross.index)

df = alpha_net.loc[both].join(
    alpha_gross.loc[both].drop(columns=["N"], errors="ignore"),
    how="inner"
)

df["alpha_gap_annual"] = df["alpha_gross_annual"] - df["alpha_net_annual"]
df["net_sig_5pct"] = df["alpha_net_p"] < 0.05
df["gross_sig_5pct"] = df["alpha_gross_p"] < 0.05

# ---------- fund types ----------
fund_types = load_fund_types_auto(funds_path)  # från Info-sheet
df = df.join(fund_types, how="left")
df["Type_bucket"] = df["Type"].apply(type_bucket)

# ---------- tables ----------
cols_order = [
    "Type", "Type_bucket", "N",
    "alpha_net_annual", "alpha_net_t", "alpha_net_p",
    "alpha_gross_annual", "alpha_gross_p",
    "alpha_gap_annual",
    "beta_net", "R2_net",
    "net_sig_5pct", "gross_sig_5pct"
]
final_table = df[cols_order].sort_values("alpha_net_annual", ascending=False)

small_table = df[[
    "Type_bucket",
    "N",
    "alpha_net_annual",
    "alpha_gross_annual",
    "alpha_gap_annual",
    "beta_net",
    "R2_net",
    "net_sig_5pct",
    "gross_sig_5pct"
]].copy()

small_table = small_table.rename(columns={
    "Type_bucket": "Type",
    "N": "Obs",
    "alpha_net_annual": "Alpha Net (%)",
    "alpha_gross_annual": "Alpha Gross (%)",
    "alpha_gap_annual": "Gross–Net Gap (%)",
    "beta_net": "Beta",
    "R2_net": "R²",
    "net_sig_5pct": "Net sig (5%)",
    "gross_sig_5pct": "Gross sig (5%)"
})

for c in ["Alpha Net (%)", "Alpha Gross (%)", "Gross–Net Gap (%)"]:
    small_table[c] *= 100

small_table = small_table.round({
    "Alpha Net (%)": 3,
    "Alpha Gross (%)": 3,
    "Gross–Net Gap (%)": 3,
    "Beta": 3,
    "R²": 3
})

display(small_table)
```

```python
min_obs = 24
start_date = "2016-01-01"

def to_period(df, date_col=None):
    if date_col:
        df[date_col] = pd.to_datetime(df[date_col], errors="coerce")
        df = df.dropna(subset=[date_col]).set_index(date_col)
    else:
        df.index = pd.to_datetime(df.index, errors="coerce")
    df.index = df.index.to_period("M")
    return df.sort_index()

def comma_to_float(df):
    for c in df.columns:
        df[c] = pd.to_numeric(df[c].astype(str).str.replace(",", ".", regex=False), errors="coerce")
    return df

def load_fund(sheet_name):
    df = pd.read_excel(funds_path, sheet_name=sheet_name)
    df.columns = df.columns.str.strip()
    if {"Year","Month"}.issubset(df.columns):
        df["Date"] = pd.to_datetime(df["Year"].astype(int).astype(str) + "-" + df["Month"].astype(int).astype(str).str.zfill(2))
        df = df.drop(columns=["Year","Month"])
    df = to_period(df, date_col="Date" if "Date" in df.columns else None)
    df = comma_to_float(df).dropna(axis=1, how="all")
    return df

def capm(funds, factors, label="net"):
    X = sm.add_constant(factors["MKT"])
    rows = []
    for fund in funds.columns:
        y = funds[fund] - factors["RF"]
        tmp = pd.concat([y.rename("y"), X], axis=1).dropna()
        if len(tmp) < min_obs:
            continue
        res = sm.OLS(tmp["y"], tmp[["const","MKT"]]).fit()
        rows.append({
            "Fund": fund,
            f"alpha_{label}_monthly": res.params["const"],
            f"alpha_{label}_annual": res.params["const"]*12,
            f"alpha_{label}_t": res.tvalues["const"],
            f"alpha_{label}_p": res.pvalues["const"],
            f"beta_{label}": res.params["MKT"],
            f"R2_{label}": res.rsquared,
            "N": int(res.nobs)
        })
    return pd.DataFrame(rows).set_index("Fund")

funds_net = load_fund("Net")
funds_gross = load_fund("Gross")
factors_subset = to_period(factors[["RF","MKT"]]).dropna()
common_months = funds_net.index.intersection(funds_gross.index).intersection(factors_subset.index)
funds_net = funds_net.loc[common_months]
funds_gross = funds_gross.loc[common_months]
factors_subset = factors_subset.loc[common_months]

alpha_net = capm(funds_net, factors_subset, "net")
alpha_gross = capm(funds_gross, factors_subset, "gross")

both = alpha_net.index.intersection(alpha_gross.index)
df = alpha_net.loc[both].join(
    alpha_gross.loc[both].drop(columns=["N"]),
    how="inner"
)
df["alpha_gap_annual"] = df["alpha_gross_annual"] - df["alpha_net_annual"]
df["net_sig_5pct"] = df["alpha_net_p"] < 0.05
df["gross_sig_5pct"] = df["alpha_gross_p"] < 0.05

fund_types = load_fund_types_auto(funds_path)  # auto-detect "Fund name"/"fund type"
df = df.join(fund_types, how="left")
df["Type_bucket"] = df["Type"].apply(type_bucket)

cols_order = [
    "Type","Type_bucket","N",
    "alpha_net_annual","alpha_net_t","alpha_net_p",
    "alpha_gross_annual","alpha_gross_p",
    "alpha_gap_annual",
    "beta_net","R2_net",
    "net_sig_5pct","gross_sig_5pct"
]
final_table = df[cols_order].sort_values("alpha_net_annual", ascending=False)

small_table = df[[
    "Type_bucket",
    "N",
    "alpha_net_annual",
    "alpha_gross_annual",
    "alpha_gap_annual",
    "beta_net",
    "R2_net",
    "net_sig_5pct",
    "gross_sig_5pct"
]].copy()

small_table = small_table.rename(columns={
    "Type_bucket": "Type",
    "N": "Obs",
    "alpha_net_annual": "Alpha Net (%)",
    "alpha_gross_annual": "Alpha Gross (%)",
    "alpha_gap_annual": "Gross–Net Gap (%)",
    "beta_net": "Beta",
    "R2_net": "R²",
    "net_sig_5pct": "Net sig (5%)",
    "gross_sig_5pct": "Gross sig (5%)"
})

for c in ["Alpha Net (%)", "Alpha Gross (%)", "Gross–Net Gap (%)"]:
    small_table[c] *= 100

small_table = small_table.round({
    "Alpha Net (%)": 3,
    "Alpha Gross (%)": 3,
    "Gross–Net Gap (%)": 3,
    "Beta": 3,
    "R²": 3
})

display(small_table)
```

Overall, the results show that fees materially reduce performance, especially for active funds. While some funds generate positive abnormal returns before fees, much of this value is absorbed by costs, and statistically significant outperformance after fees is rare.

```python
chosen_stocks = ["Volvo AB A", "Assa Abloy AB B", "Handelsbanken AB A", "Industrivarden AB A"]
mkt = "market portfolio"

xidx = returns.index.intersection(factors.index)
re = returns.loc[xidx, chosen_stocks].apply(pd.to_numeric, errors="coerce")
fas = factors.loc[xidx, ["rf", "rm_rf"]].apply(pd.to_numeric, errors="coerce")

#print("overlap months:", len(xidx))
#print("missing stocks columns:", [s for s in chosen_stocks if s not in returns.columns])

rows = []
for s in chosen_stocks:
    yy = (re[s] - fas["rf"]).rename("excess")
    xx = sm.add_constant(fas["M"])
    d_f = pd.concat([yy, xx], axis=1).dropna()
    k = sm.OLS(d_f["excess"], d_f[["const", "rm_rf"]]).fit()

    rows.append({ 
        "stocks": s, 
        "beta": k.params["rm_rf"],
        "monthly_alpha": k.params["const"],
        "sigma": k.mse_resid,
        "nobs": int(k.nobs),
     })

estimates = pd.DataFrame(rows).set_index("stocks")
display(estimates.round(6))

mu_e = fas["rm_rf"].dropna().mean()
sig = fas["rm_rf"].dropna().var(ddof=1)

muu = pd.concat([
    pd.Series({mkt: mu_e}),
    (estimates["monthly_alpha"] + estimates["beta"]* mu_e)
])

beta_e = pd.concat([
    pd.Series({mkt: 1.0}),
    (estimates["beta"])
])

res_e = pd.concat([
    pd.Series({mkt: 0.0}),
    (estimates["sigma"])
])

sigmas = np.outer(beta_e.values, beta_e.values) * sig + np.diag(res_e.values)
sigmas = pd.DataFrame(sigmas, index=muu.index, columns=muu.index)

weight = np.linalg.solve(sigmas.values, muu.values)
w = pd.Series(weight, index=muu.index)
w = w / w.sum()

#weights = (w * 100).rename("weights_%")
display((w*100).rename("weights_%").to_frame().round(6))

print("Sum of weights:", w.sum().round(6))
print("Beta of the portfolio:", float((w * beta_e).sum()))
```
